# Aula 01 - Bioinformática: quando a Biologia vira dado

Nesta prática vamos tratar sequências de DNA como dados computacionais. As sequências são **sintéticas e didáticas**.

## 1. Uma sequência pode ser tratada como uma string
A primeira ideia computacional é simples: uma sequência de DNA pode ser representada por caracteres A, T, C e G.

```python
sequencia = 'ATGCGTACGTTAGCGGCTAA'
print('Sequência:', sequencia)
print('Tamanho:', len(sequencia))
for base in 'ATCG':
    print(base, sequencia.count(base))
```

## 2. Transformando sequência em medida
Vamos calcular o conteúdo GC, isto é, a proporção de G e C na sequência.

```python
gc = 100 * (sequencia.count('G') + sequencia.count('C')) / len(sequencia)
print(f'Conteúdo GC: {gc:.2f}%')
```

## 3. Criando uma pequena base de sequências
Agora passamos de uma única string para uma tabela, aproximando a atividade de uma análise de dados.

```python
import pandas as pd
import matplotlib.pyplot as plt

dados = {
    'amostra': ['Amostra_A','Amostra_B','Amostra_C','Amostra_D','Amostra_E','Amostra_F'],
    'sequencia': [
        'ATGCGTACGTTAGCGGCTAA',
        'ATATATGCGTATATGCAAAT',
        'GCGCGCGTACGCGGCCGTAA',
        'ATGCCCTTAGGCTAACCGTA',
        'TTATTAACGATATTAACGTA',
        'CCGGCGGCCGATCGGCGCTA'
    ]
}

df = pd.DataFrame(dados)
df['tamanho'] = df['sequencia'].str.len()
df['gc_percentual'] = df['sequencia'].apply(lambda s: 100*(s.count('G')+s.count('C'))/len(s))
df
```

## 4. Visualizando o resultado
Qual amostra apresenta maior conteúdo GC?

```python
df.plot(x='amostra', y='gc_percentual', kind='bar', legend=False)
plt.title('Conteúdo GC das sequências didáticas')
plt.xlabel('Amostra')
plt.ylabel('GC (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
