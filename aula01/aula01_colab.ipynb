# Aula 01 - Bioinformática: quando a Biologia vira dado

Nesta prática vamos tratar sequências de DNA como dados computacionais. As sequências são **sintéticas e didáticas**.

> **Importante:** você não precisa saber Python. Cada comando será explicado antes e dentro do código.

## 1. Uma sequência pode ser tratada como uma string

Em Python, um texto é chamado de **string**. Como uma sequência de DNA pode ser escrita usando letras A, T, C e G, podemos armazená-la em uma variável de texto.

Vamos usar três recursos:
- `=` guarda um valor em uma variável;
- `print()` mostra uma informação na tela;
- `len()` conta quantos caracteres existem em um texto.

```python
# Criamos uma variável chamada 'sequencia'.
# O sinal = significa: guarde o valor que está à direita dentro do nome que está à esquerda.
sequencia = 'ATGCGTACGTTAGCGGCTAA'

# print() mostra um resultado na tela.
print('Sequência:', sequencia)

# len() significa length, isto é, comprimento.
# Ele conta quantos caracteres existem dentro da sequência.
print('Tamanho:', len(sequencia))
```

## 2. Contando A, T, C e G

O método `.count()` conta quantas vezes um caractere aparece em uma string.

Primeiro faremos da forma mais simples, repetindo o comando. Isso facilita a compreensão de quem está começando.

```python
# sequencia.count('A') procura a letra A e conta quantas vezes ela aparece.
print('Quantidade de A:', sequencia.count('A'))

# Fazemos a mesma operação para as outras bases.
print('Quantidade de T:', sequencia.count('T'))
print('Quantidade de C:', sequencia.count('C'))
print('Quantidade de G:', sequencia.count('G'))
```

### Uma versão mais automática

Depois de entender a contagem manual, podemos usar `for`. O `for` é uma estrutura de repetição: ele executa o mesmo comando para cada elemento de uma coleção.

```python
# A expressão 'ATCG' contém as quatro bases que queremos analisar.
# O for pega uma letra de cada vez e coloca temporariamente dentro da variável 'base'.
for base in 'ATCG':
    # Esta linha está recuada porque pertence ao bloco do for.
    # Ela será executada quatro vezes: uma para A, uma para T, uma para C e uma para G.
    print(base, sequencia.count(base))
```

## 3. Transformando sequência em uma medida: conteúdo GC

O **conteúdo GC** é a porcentagem de bases G e C presentes na sequência.

Nesta aula, o objetivo principal não é interpretar biologicamente o valor. Queremos observar como uma sequência de letras pode ser transformada em uma variável numérica.

```python
# Primeiro contamos G.
quantidade_g = sequencia.count('G')

# Depois contamos C.
quantidade_c = sequencia.count('C')

# Somamos G + C.
total_gc = quantidade_g + quantidade_c

# Dividimos a quantidade de G+C pelo tamanho total da sequência.
# Multiplicamos por 100 para transformar o resultado em porcentagem.
gc_percentual = 100 * total_gc / len(sequencia)

# :.2f pede ao Python para mostrar o número com duas casas decimais.
print(f'Conteúdo GC: {gc_percentual:.2f}%')
```

## 4. Passando de uma sequência para uma pequena base de dados

Agora analisaremos várias sequências. Para isso, usaremos a biblioteca **pandas**, muito utilizada em Ciência de Dados para trabalhar com tabelas.

Também usaremos **matplotlib** para criar um gráfico.

`import` significa que estamos trazendo uma biblioteca para dentro do nosso programa.

```python
# Importa a biblioteca pandas e dá a ela o apelido pd.
import pandas as pd

# Importa a parte pyplot da biblioteca matplotlib e dá a ela o apelido plt.
import matplotlib.pyplot as plt
```

### Criando os dados

Vamos criar uma estrutura chamada **dicionário**. Ela associa um nome de campo a uma lista de valores. Depois o pandas transformará essa estrutura em uma tabela.

```python
# Criamos um dicionário chamado dados.
# A chave 'amostra' contém os nomes das seis amostras.
# A chave 'sequencia' contém as respectivas sequências sintéticas.
dados = {
    'amostra': ['Amostra_A', 'Amostra_B', 'Amostra_C', 'Amostra_D', 'Amostra_E', 'Amostra_F'],
    'sequencia': [
        'ATGCGTACGTTAGCGGCTAA',
        'ATATATGCGTATATGCAAAT',
        'GCGCGCGTACGCGGCCGTAA',
        'ATGCCCTTAGGCTAACCGTA',
        'TTATTAACGATATTAACGTA',
        'CCGGCGGCCGATCGGCGCTA'
    ]
}

# pd.DataFrame() transforma o dicionário em uma tabela.
# DataFrame é o nome dado pelo pandas a uma estrutura tabular com linhas e colunas.
df = pd.DataFrame(dados)

# Apenas escrever o nome da variável no final da célula faz o Colab exibir a tabela.
df
```

## 5. Criando novas informações a partir dos dados

Agora criaremos duas novas colunas: o tamanho de cada sequência e seu conteúdo GC.

A lógica é a mesma que usamos com uma única sequência, mas agora será aplicada a todas as linhas da tabela.

```python
# df['sequencia'] seleciona a coluna chamada sequencia.
# .str indica que queremos usar operações de texto nessa coluna.
# .len() calcula o tamanho de cada sequência.
# O resultado é armazenado em uma nova coluna chamada tamanho.
df['tamanho'] = df['sequencia'].str.len()

# Criamos uma função simples para calcular o conteúdo GC de UMA sequência.
def calcular_gc(s):
    # Conta G e C, soma os dois valores, divide pelo tamanho e multiplica por 100.
    return 100 * (s.count('G') + s.count('C')) / len(s)

# .apply() aplica a função calcular_gc a cada sequência existente na coluna.
# O resultado é guardado na nova coluna gc_percentual.
df['gc_percentual'] = df['sequencia'].apply(calcular_gc)

# Exibe a tabela atualizada.
df
```

## 6. Visualizando o resultado

Um gráfico facilita a comparação entre as amostras. Observe qual barra é mais alta e qual é mais baixa.

```python
# df.plot() pede ao pandas para criar um gráfico usando os dados da tabela.
# x define a coluna usada no eixo horizontal.
# y define a coluna usada no eixo vertical.
# kind='bar' escolhe um gráfico de barras.
df.plot(x='amostra', y='gc_percentual', kind='bar', legend=False)

# Define o título do gráfico.
plt.title('Conteúdo GC das sequências didáticas')

# Define os nomes dos eixos.
plt.xlabel('Amostra')
plt.ylabel('GC (%)')

# Gira os nomes das amostras para facilitar a leitura.
plt.xticks(rotation=45)

# Ajusta automaticamente os espaços do gráfico.
plt.tight_layout()

# Mostra o gráfico na tela.
plt.show()
```
